# Project Sentinel: synthetic payment intelligence
All records are fictional. Execute from the repository environment. The analysis is descriptive, not a real fraud or AML assessment.

```python
from pathlib import Path
import sys
root = Path.cwd()
if not (root / 'sentinel').exists():
    root = root.parent
sys.path.insert(0, str(root))
from sentinel.generate import generate
from sentinel.analysis import validate, providers, daily, reconciliation
from sentinel.model import evaluate
import matplotlib.pyplot as plt
import pandas as pd
```

## 1. Reproducible dataset and validation

```python
df = generate(100_000, 42)
validate(df)
print(df.shape)
print(df.status.value_counts())
df.head()
```

## 2. Provider performance
Rates use all attempts as denominator. Amounts are gross attempted KES.

```python
p = providers(df)
p
```

```python
p.set_index('provider')[['approval_rate','failure_rate','pending_rate']].plot.bar(figsize=(9,4))
plt.ylabel('Fraction of attempts'); plt.tight_layout(); plt.show()
```

## 3. Daily activity and reconciliation

```python
daily(df).set_index('day').attempts.plot(figsize=(10,4), title='Daily transaction attempts')
plt.show()
r = reconciliation(df)
print('Exceptions:',len(r),'Absolute difference KES:',r.difference_kes.abs().sum())
r.head()
```

## 4. Temporal holdout model evaluation
Compare average precision against label prevalence. Synthetic labels are not actual fraud outcomes.

```python
metrics = evaluate(df)
pd.Series(metrics)
```

## 5. Limitations
Provider behavior and labels are determined by simulation assumptions. No causal or real-world fraud conclusions follow. Further real-world validation and governance are required.
